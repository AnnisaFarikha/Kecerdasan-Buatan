#Central Tendency

```python
# A. Mendapatkan Nilai Mean
import numpy as np
score = [90, 80, 70, 65, 110, 80, 83, 67, 94]
x = np.mean(score)
print(x)

# B. Mendapatkan Nilai Median
import numpy as np
score = [90, 80, 70, 65, 110, 80, 83, 67, 94]
x = np.median(score)
print(x)

# C. Mendapatkan Nilai Modus / Mode
from scipy import stats
score = [90, 80, 70, 65, 110, 80, 83, 67, 94]
x = stats.mode(score, keepdims=True)
print(x)
```

#Spread

```python
# A. Mendapatkan Nilai Range
import numpy as np
data = [2, 3, 7, 8, 9, 12, 15, 20, 50, 42]
minimum = np.min(data)
maximum = np.max(data)
x = maximum - minimum
print(x)

# B. Mendapatkan Nilai Percentile
import numpy as np
data = [90, 80, 70, 65, 110, 80, 83, 67, 94]
x = np.percentile(data, 90)
print(x)

# C. Mendapatkan Nilai Quartile dan Interquartile Range
import numpy as np
data = [3, 4, 4, 5, 5, 6, 6, 7, 8, 8, 9, 10]
Q1 = np.median(data[:int(len(data)/2)])
Q3 = np.median(data[int(len(data)/2):])
IQR = Q3 - Q1
print(Q1, Q3, IQR)

# D. Mendapatkan Nilai Variance dan Standart Deviation
import numpy as np
data= [86, 87, 88, 86, 87, 85, 86]
mean = np.mean(data)
var = np.var(data)
std = np.std(data)
print(mean, var, std)
```

#Tugas 1: Modus

```python
from scipy import stats

data = [1,2,3,4,5,5,6,2,3,4,5,6,6,4,3,2,1,4,3,5,6,6,5,4,5]

modus = stats.mode(data, keepdims=True)

print("Data =", data)
print("Modus =", modus.mode[0])
print("Frekuensi =", modus.count[0])
```

#Tugas 2: Percentile

```python
import numpy as np

data = [9,10,11,6,8,7,7,5,4,5]

p25 = np.percentile(data, 25)
p75 = np.percentile(data, 75)

print("Data =", data)
print("Persentil 25 =", p25)
print("Persentil 75 =", p75)
```

#Tugas 3: Hasil Ujian

```python
import numpy as np

data = [50,54,62,50,52,59,61,63,65,10,53,63,65,50,59,62,50,51,57,60,63,65,65,53,99]

# A. Urutkan data
urut = sorted(data)

# B. Q1, Q3, IQR
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1

# C. Upper dan Lower Bound
lower = Q1 - (1.5 * IQR)
upper = Q3 + (1.5 * IQR)

# D. Outlier
outlier = []
for x in data:
    if x < lower or x > upper:
        outlier.append(x)

print("Data Urut =", urut)
print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
print("Lower Bound =", lower)
print("Upper Bound =", upper)
print("Outlier =", outlier)
```

#Tugas 4: Standart Deviasi

```python
import numpy as np

# titik tengah kelas
x = np.array([43,48,53,58,63,68])

# frekuensi
f = np.array([2,3,6,4,5,2])

# membuat data sesuai frekuensi
data = np.repeat(x, f)

std = np.std(data)

print("Data =", data)
print("Standar Deviasi =", round(std, 2))
```
