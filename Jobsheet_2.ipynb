#Operasi Matematika Menggunakan Python

```python
# A. Mendapatkan nilai minimum dan maksimum
x = min(5, 10, 25)
y = max(6, 12, 18)
print(x)
print(y)

# B. Mendapatkan nilai absolut
x = abs(-7.25)
print(x)

# C. Mendapatkan nilai pangkat
import math
x = math.pow(4, 3)
print(x)

# D. Mendapatkan nilai akar
import math
x = math.sqrt(64)
print(x)

# E. Mendapatkan nilai pembulatan
import math
x = math.ceil(1.4)
y = math.floor(1.4)
print(x)
print(y)

# F. Mendapatkan nilai konstanta (PI)
import math
x = math.pi
print(x)
```

#Menggunakan Fungsi MATH

```python
import math

print("math.sin(30) =", math.sin(30))
print("math.cos(30) =", math.cos(30))
print("math.tan(30) =", math.tan(30))

print("math.log(10) =", math.log(10))
print("math.log10(100) =", math.log10(100))

print("math.degrees(3.14) =", math.degrees(3.14))

print("math.dist([2,3], [5,7]) =", math.dist([2,3], [5,7]))

print("math.factorial(5) =", math.factorial(5))

print("math.fmod(10,3) =", math.fmod(10,3))

print("math.isnan(5.6) =", math.isnan(5.6))

print("math.gamma(5) =", math.gamma(5))

print("math.perm(5,2) =", math.perm(5,2))
```

#Operasi Matriks dan Vektor dengan Python

```python
# A. Membuat matriks di pyhon
import numpy as np
A = np.array([[80, 30, 25],
[50, 20, 10]])
B = np.array([[30, 60, 15],
[10, 30, 30]])

hasil = A + B
print(hasil)

# B. Melakukan operasi dot product
import numpy as np
A = np.array([10, 20, 15])
B = np.array([16, 67, 35])

hasil = np.dot(A, B)
print(hasil)

# C. Melakukan operasi cross product
import numpy as np
A = np.array([10, 20, 15])
B = np.array([16, 67, 35])

hasil = np.cross(A, B)
print(hasil)
```

#Tugas 1: Perkalian

```python
import numpy as np

A = np.array([[3, 1],
              [2, 0]])

B = np.array([[2, 1],
              [1, -1]])

C = np.array([[-1, 1, 2],
              [2, 0, -1]])

hasil1 = np.dot(A, B)   # A x B
hasil2 = np.dot(B, C)   # B . C

print("A x B =")
print(hasil1)

print("B . C =")
print(hasil2)
```

#Tugas 2: Faktorial

```python
import math

a = math.factorial(6)                       # 6!
b = math.factorial(8) / math.factorial(5)   # 8! / 5!

print("6! =", a)
print("8! / 5! =", b)
```

#Tugas 3: Vektor

```python
import numpy as np

A = np.array([5, 3, 7])
B = np.array([12, -3, 1])

# Panjang vektor
panjang_A = np.linalg.norm(A)
panjang_B = np.linalg.norm(B)

dot = np.dot(A, B)       # Dot Product
cross = np.cross(A, B)   # Cross Product

print("Panjang Vektor A =", panjang_A)
print("Panjang Vektor B =", panjang_B)

print("A . B =", dot)

print("A x B =")
print(cross)
```
